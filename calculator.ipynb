```python
print("easy calculator")
print("choose operation only:+,-,*,/")
operation = input("choose your symbol  : +, -, *, /")
first = int(input("enter your first  number :"))
second = int(input("enter your second  number :"))

if operation =="+":
    add = first + second
    print ("result:",add)
    
elif operation =="-":
    sub = first - second
    print ("result:",sub)
elif operation =="*":
    mult = first * second
    print ("result:",mult)
elif operation =="/":
    if second ==0:
        print("cannot divide by zero")
    else:
        result = first/second
        print("result:",result)

else:
    print("invalid operator")
```
