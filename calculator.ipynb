### arithmetic operators
* '+'
* '-'
* '*'
* '/'
* '//'
* '%'
* '**'

```python
print("calculator")
operator = input("choose operator : +, -, *, /, //, %, ** :")
first = float(input("enter your first number :"))
second = float(input("enter your second number :"))
if operator =="+":
    result = first + second
    print("result:",result)
elif operator =="-":
    result = first - second
    print("result:",result)
elif operator =="*":
    result = first - second
    print("result:",result)
elif operator =="/":
    result = first / second
    print("result:",result)
elif operator =="//":
    result = first // second
    print("result:",result)
elif operator =="%":
    result = first % second
    print("result:",result)
elif operator =="**":
    result = first ** second
    print("result:",result)
else:
    print("invalid operator")
```
