### 3.Factorial Calculation Write a program using a loop to calculate the factorial of a given number n.

```python
n = int(input("Enter a number: "))
factorial = 1
for i in range(1, n + 1):
    factorial = factorial * i
print("Factorial =", factorial)
```

### 5 Reverse String Use a while loop to reverse the string "assignment" without using slicing.


```python
name = "assignment"
rev = ""
i = len(name)-1
while i >= 0:
    rev = rev + name[i]
    i = i -1
print(f'reverse of the string "assignment is {rev}')
```

### 8.Palindrome Concept
A palindrome is a word, phrase, or number that reads the same backward as forward. Examples: "madam", "racecar", 121.

```python
text = input("Enter a word or number: ")
rev = ""
for i in text:
    rev = i + rev

if text == rev:
    print(text, "is a palindrome")
else:
    print(text, "is not a palindrome")
```

### 9.Armstong Number

```python
num = int(input("Enter a number: "))
org_num = num
total = 0
while num > 0:
    digit = num % 10
    total = total + digit ** 3
    num = num // 10
if total == org_num:
    print("YES this is an Armstrong number")
else:
    print("NO this is not an Armstrong number ")
```

### 10.Print all prime numbers between 1 and 100.

```python
print("All prime numbers between 1 and 100")
for num in range(1,101):
    count=0
    for i in range(1, num+1):
        if num % i == 0:
            count = count + 1
    if count == 2:
        print(num)
```
